## 1. Implementación de los tres métodos


```python
import random
import time

def busqueda_secuencial(lista, objetivo):
    for i, e in enumerate(lista):
        if e == objetivo:
            return i
    return -1

def busqueda_binaria(lista, objetivo):
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izq = medio + 1
        else:
            der = medio - 1
    return -1

class TablaHash:
    def __init__(self, capacidad=10007):
        self.capacidad = capacidad
        self.tabla = [[] for _ in range(capacidad)]
    
    def _hash(self, clave):
        return hash(clave) % self.capacidad
    
    def insertar(self, clave):
        indice = self._hash(clave)
        if clave not in self.tabla[indice]:
            self.tabla[indice].append(clave)
    
    def buscar(self, clave):
        indice = self._hash(clave)
        return clave in self.tabla[indice]

# Prueba rápida
datos = sorted(random.sample(range(100000), 1000))
print("Secuencial 42:", busqueda_secuencial(datos, datos[42]))
print("Binaria 42:   ", busqueda_binaria(datos, datos[42]))
th = TablaHash()
for x in datos:
    th.insertar(x)
print("Hash 42:      ", th.buscar(datos[42]))
```

## 2. Comparativa experimental


```python
def medir_busquedas(n, num_consultas=200):
    datos_lista = list(range(n))
    random.shuffle(datos_lista)
    datos_ordenados = sorted(datos_lista)
    
    th = TablaHash(capacidad=max(n * 2, 10007))
    for x in datos_lista:
        th.insertar(x)
    
    consultas = random.sample(range(n * 2), num_consultas)
    
    # Secuencial
    t0 = time.perf_counter()
    for c in consultas:
        busqueda_secuencial(datos_lista, c)
    t_sec = time.perf_counter() - t0
    
    # Binaria
    t0 = time.perf_counter()
    for c in consultas:
        busqueda_binaria(datos_ordenados, c)
    t_bin = time.perf_counter() - t0
    
    # Hash
    t0 = time.perf_counter()
    for c in consultas:
        th.buscar(c)
    t_hash = time.perf_counter() - t0
    
    return t_sec, t_bin, t_hash

print(f"{'n':>8} {'Secuencial':>12} {'Binaria':>12} {'Hash':>12}")
print("-" * 50)
for n in [1000, 10000, 50000]:
    ts, tb, th = medir_busquedas(n)
    print(f"{n:8} {ts:12.6f} {tb:12.6f} {th:12.6f}")
```

## 3. Cuadros de complejidad

| Método      | Tiempo promedio | Tiempo peor caso | Espacio extra |
|-------------|-----------------|------------------|---------------|
| Secuencial  | O(n)            | O(n)             | O(1)          |
| Binaria     | O(log n)        | O(log n)         | O(1)          |
| Hash        | O(1)            | O(n)             | O(n)          |


## 4. Tu propio método de búsqueda

Diseña e implementa un método propio (puedes combinar ideas: por ejemplo, búsqueda por saltos, interpolación, o un híbrido).


```python
def busqueda_propia(lista, objetivo):
    """
    Ejemplo: Jump Search (búsqueda por saltos)
    Requiere lista ordenada. Salta de bloque en bloque y luego busca linealmente.
    """
    import math
    n = len(lista)
    paso = int(math.sqrt(n))
    prev = 0
    
    while prev < n and lista[min(paso, n) - 1] < objetivo:
        prev = paso
        paso += int(math.sqrt(n))
        if prev >= n:
            return -1
    
    for i in range(prev, min(paso, n)):
        if lista[i] == objetivo:
            return i
    return -1

# Prueba
ordenada = list(range(0, 1000, 3))
print("Jump search de 300:", busqueda_propia(ordenada, 300))
print("Jump search de 301:", busqueda_propia(ordenada, 301))
```
