## 1. Torres de Hanoi – contar movimientos


```python
def hanoi(n, origen="A", auxiliar="B", destino="C", contador=None):
    if contador is None:
        contador = [0]
    if n == 1:
        # print(f"Mover disco 1 de {origen} a {destino}")
        contador[0] += 1
        return contador[0]
    hanoi(n - 1, origen, destino, auxiliar, contador)
    # print(f"Mover disco {n} de {origen} a {destino}")
    contador[0] += 1
    hanoi(n - 1, auxiliar, origen, destino, contador)
    return contador[0]

for discos in range(1, 9):
    movimientos = hanoi(discos)
    teorico = 2 ** discos - 1
    print(f"n={discos}: movimientos={movimientos}, teórico={teorico}, ¿correcto? {movimientos == teorico}")
```

## 2. Convertir algoritmos iterativos a recursivos


```python
# a) Búsqueda lineal recursiva
def busqueda_lineal_rec(lista, objetivo, indice=0):
    if indice >= len(lista):
        return -1
    if lista[indice] == objetivo:
        return indice
    return busqueda_lineal_rec(lista, objetivo, indice + 1)

# b) Máximo de una lista recursivo
def maximo_rec(lista):
    if len(lista) == 1:
        return lista[0]
    resto_max = maximo_rec(lista[1:])
    return lista[0] if lista[0] > resto_max else resto_max

# c) Contar vocales recursivo
def contar_vocales(texto):
    if not texto:
        return 0
    es_vocal = 1 if texto[0].lower() in "aeiouáéíóú" else 0
    return es_vocal + contar_vocales(texto[1:])

# Pruebas
datos = [4, 7, 1, 9, 3, 8]
print("Búsqueda de 9:", busqueda_lineal_rec(datos, 9))
print("Máximo:", maximo_rec(datos))
print("Vocales en 'Programación':", contar_vocales("Programación"))
```

## 3. Fibonacci ingenuo vs memoizado – comparación de tiempos


```python
import time
from functools import lru_cache

def fib_ingenuo(n):
    if n < 2:
        return n
    return fib_ingenuo(n - 1) + fib_ingenuo(n - 2)

@lru_cache(maxsize=None)
def fib_memo(n):
    if n < 2:
        return n
    return fib_memo(n - 1) + fib_memo(n - 2)

for n in [30, 35]:
    t0 = time.perf_counter()
    r1 = fib_ingenuo(n)
    t1 = time.perf_counter() - t0
    
    t0 = time.perf_counter()
    r2 = fib_memo(n)
    t2 = time.perf_counter() - t0
    
    print(f"n={n}: ingenuo={t1:.4f}s, memoizado={t2:.6f}s, resultado={r1}")
```
