## 1. Implementación de los algoritmos internos


```python
import random
import time
import copy

def burbuja(lista):
    n = len(lista)
    for i in range(n):
        intercambio = False
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                intercambio = True
        if not intercambio:
            break
    return lista

def quicksort(lista):
    if len(lista) <= 1:
        return lista
    pivote = lista[len(lista) // 2]
    menores = [x for x in lista if x < pivote]
    iguales = [x for x in lista if x == pivote]
    mayores = [x for x in lista if x > pivote]
    return quicksort(menores) + iguales + quicksort(mayores)

def shell_sort(lista):
    n = len(lista)
    gap = n // 2
    while gap > 0:
        for i in range(gap, n):
            temp = lista[i]
            j = i
            while j >= gap and lista[j - gap] > temp:
                lista[j] = lista[j - gap]
                j -= gap
            lista[j] = temp
        gap //= 2
    return lista

def radix_sort(lista):
    if not lista:
        return lista
    maximo = max(lista)
    exp = 1
    while maximo // exp > 0:
        n = len(lista)
        salida = [0] * n
        conteo = [0] * 10
        for i in range(n):
            indice = (lista[i] // exp) % 10
            conteo[indice] += 1
        for i in range(1, 10):
            conteo[i] += conteo[i - 1]
        for i in range(n - 1, -1, -1):
            indice = (lista[i] // exp) % 10
            salida[conteo[indice] - 1] = lista[i]
            conteo[indice] -= 1
        lista[:] = salida
        exp *= 10
    return lista

# Verificación rápida
prueba = [64, 34, 25, 12, 22, 11, 90]
print("Burbuja: ", burbuja(prueba[:]))
print("Quick:   ", quicksort(prueba[:]))
print("Shell:   ", shell_sort(prueba[:]))
print("Radix:   ", radix_sort(prueba[:]))
```

## 2. Comparativa de tiempos


```python
def medir(func, datos):
    copia = copy.deepcopy(datos)
    t0 = time.perf_counter()
    func(copia)
    return time.perf_counter() - t0

algoritmos = {
    "Burbuja": burbuja,
    "Quicksort": quicksort,
    "Shell": shell_sort,
    "Radix": radix_sort,
}

tamanos = [2000, 8000]
tipos = {
    "Aleatorio": lambda n: [random.randint(0, 10000) for _ in range(n)],
    "Ordenado":  lambda n: list(range(n)),
    "Inverso":   lambda n: list(range(n, 0, -1)),
}

print(f"{'Algoritmo':12} {'Tipo':12} {'n':6} {'Tiempo (s)':12}")
print("-" * 50)

for n in tamanos:
    for tipo_nombre, generador in tipos.items():
        base = generador(n)
        for nombre, func in algoritmos.items():
            # Burbuja es muy lento en 8000, lo saltamos si es necesario
            if nombre == "Burbuja" and n >= 8000:
                print(f"{nombre:12} {tipo_nombre:12} {n:6} {'(omitido)':12}")
                continue
            t = medir(func, base)
            print(f"{nombre:12} {tipo_nombre:12} {n:6} {t:12.6f}")
```

## 3. Ordenamiento externo – Intercalación y Mezcla Directa


```python
def intercalacion(a, b):
    resultado = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            resultado.append(a[i])
            i += 1
        else:
            resultado.append(b[j])
            j += 1
    resultado.extend(a[i:])
    resultado.extend(b[j:])
    return resultado

def mezcla_directa(lista):
    n = len(lista)
    ancho = 1
    while ancho < n:
        for i in range(0, n, 2 * ancho):
            izq = lista[i:i + ancho]
            der = lista[i + ancho:i + 2 * ancho]
            lista[i:i + 2 * ancho] = intercalacion(izq, der)
        ancho *= 2
    return lista

datos = [38, 27, 43, 3, 9, 82, 10, 15]
print("Original:     ", datos)
print("Mezcla directa:", mezcla_directa(datos[:]))
print("sorted():     ", sorted(datos))
print("¿Iguales?", mezcla_directa(datos[:]) == sorted(datos))
```
