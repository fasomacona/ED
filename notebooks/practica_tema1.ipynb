### 3.1 Observar el crecimiento de una lista de Python

Python redimensiona internamente la lista cuando se llena. Observemos la capacidad interna (aproximada).

```python
import sys

lista = []
capacidades = []
tamanos = []

for i in range(64):
    lista.append(i)
    # sys.getsizeof incluye overhead del objeto; sirve para ver saltos de crecimiento
    tamanos.append(len(lista))
    capacidades.append(sys.getsizeof(lista))

print("n  | sizeof (bytes)")
print("---+---------------")
prev = 0
for n, sz in zip(tamanos, capacidades):
    if sz != prev:
        print(f"{n:3d} | {sz}")
        prev = sz

print("\n💡 Observa que el tamaño en bytes crece a saltos (estrategia de redimensionamiento).")
```

```python
import time
import random

def medir_tiempo(func, lista, objetivo, repeticiones=5):
    """Retorna el tiempo promedio en milisegundos."""
    tiempos = []
    for _ in range(repeticiones):
        inicio = time.perf_counter()
        func(lista, objetivo)
        fin = time.perf_counter()
        tiempos.append((fin - inicio) * 1000)
    return sum(tiempos) / len(tiempos)


tamanos = [1_000, 10_000, 100_000, 1_000_000]
resultados = []

for n in tamanos:
    lista = list(range(n))  # ya ordenada
    objetivo = n - 1        # peor caso para lineal (está al final)

    t_lin = medir_tiempo(busqueda_lineal, lista, objetivo)
    t_bin = medir_tiempo(busqueda_binaria, lista, objetivo)

    resultados.append({
        "n": n,
        "Lineal (ms)": round(t_lin, 4),
        "Binaria (ms)": round(t_bin, 4),
        "Ratio lineal/binaria": round(t_lin / t_bin, 1) if t_bin > 0 else float("inf"),
    })

df_tiempos = pd.DataFrame(resultados)
df_tiempos
```
