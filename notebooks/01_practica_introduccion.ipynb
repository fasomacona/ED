## 1. TDA Conjunto (Set)

Implementa las operaciones: agregar, eliminar, pertenece, union, interseccion.


```python
class Conjunto:
    """TDA Conjunto implementado con una lista interna (sin usar set de Python)."""
    
    def __init__(self, elementos=None):
        self._datos = []
        if elementos:
            for e in elementos:
                self.agregar(e)
    
    def agregar(self, x):
        if x not in self._datos:
            self._datos.append(x)
    
    def eliminar(self, x):
        if x in self._datos:
            self._datos.remove(x)
    
    def pertenece(self, x):
        return x in self._datos
    
    def union(self, otro):
        resultado = Conjunto(self._datos)
        for e in otro._datos:
            resultado.agregar(e)
        return resultado
    
    def interseccion(self, otro):
        resultado = Conjunto()
        for e in self._datos:
            if otro.pertenece(e):
                resultado.agregar(e)
        return resultado
    
    def __repr__(self):
        return "{" + ", ".join(map(str, self._datos)) + "}"
    
    def __len__(self):
        return len(self._datos)

# Pruebas
a = Conjunto([1, 2, 3, 4])
b = Conjunto([3, 4, 5, 6])
print("A:", a)
print("B:", b)
print("Unión:", a.union(b))
print("Intersección:", a.interseccion(b))
print("¿5 pertenece a A?", a.pertenece(5))
a.agregar(5)
print("Después de agregar 5:", a)
```

## 2. Comparación de eficiencia: lista vs set de Python


```python
import time
import random

def medir_busqueda(estructura, claves, nombre):
    inicio = time.perf_counter()
    encontrados = sum(1 for c in claves if c in estructura)
    tiempo = time.perf_counter() - inicio
    print(f"{nombre:20s} → encontrados={encontrados}, tiempo={tiempo:.6f}s")
    return tiempo

tamanos = [1_000, 10_000, 100_000]
resultados = []

for n in tamanos:
    print(f"\n=== Tamaño n = {n} ===")
    datos = list(range(n))
    random.shuffle(datos)
    claves = random.sample(range(n * 2), 500)  # 500 búsquedas
    
    lista = datos[:]
    conjunto = set(datos)
    
    t_lista = medir_busqueda(lista, claves, "Lista (O(n))")
    t_set = medir_busqueda(conjunto, claves, "Set (O(1) promedio)")
    resultados.append((n, t_lista, t_set))
```
