## 1. Árbol Binario de Búsqueda – recorridos


```python
class NodoArbol:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

class ABB:
    def __init__(self):
        self.raiz = None
    
    def insertar(self, valor):
        self.raiz = self._insertar(self.raiz, valor)
    
    def _insertar(self, nodo, valor):
        if nodo is None:
            return NodoArbol(valor)
        if valor < nodo.valor:
            nodo.izq = self._insertar(nodo.izq, valor)
        elif valor > nodo.valor:
            nodo.der = self._insertar(nodo.der, valor)
        return nodo

def inorden(nodo, resultado=None):
    if resultado is None:
        resultado = []
    if nodo:
        inorden(nodo.izq, resultado)
        resultado.append(nodo.valor)
        inorden(nodo.der, resultado)
    return resultado

def preorden(nodo, resultado=None):
    if resultado is None:
        resultado = []
    if nodo:
        resultado.append(nodo.valor)
        preorden(nodo.izq, resultado)
        preorden(nodo.der, resultado)
    return resultado

def postorden(nodo, resultado=None):
    if resultado is None:
        resultado = []
    if nodo:
        postorden(nodo.izq, resultado)
        postorden(nodo.der, resultado)
        resultado.append(nodo.valor)
    return resultado

arbol = ABB()
for v in [50, 30, 70, 20, 40, 60, 80, 10, 25]:
    arbol.insertar(v)

print("Inorden:  ", inorden(arbol.raiz))
print("Preorden: ", preorden(arbol.raiz))
print("Postorden:", postorden(arbol.raiz))
```

## 2. Grafo ponderado + Dijkstra (camino mínimo)


```python
from collections import defaultdict
import heapq

class Grafo:
    def __init__(self):
        self.ady = defaultdict(list)
    
    def agregar_arista(self, u, v, peso):
        self.ady[u].append((v, peso))
        self.ady[v].append((u, peso))  # no dirigido
    
    def dijkstra(self, origen):
        dist = {v: float("inf") for v in self.ady}
        dist[origen] = 0
        predecesor = {v: None for v in self.ady}
        pq = [(0, origen)]
        
        while pq:
            d_act, u = heapq.heappop(pq)
            if d_act > dist[u]:
                continue
            for v, peso in self.ady[u]:
                nueva = d_act + peso
                if nueva < dist[v]:
                    dist[v] = nueva
                    predecesor[v] = u
                    heapq.heappush(pq, (nueva, v))
        return dist, predecesor
    
    def camino(self, predecesor, destino):
        ruta = []
        actual = destino
        while actual is not None:
            ruta.append(actual)
            actual = predecesor[actual]
        return list(reversed(ruta))

# Ejemplo: ciudades
g = Grafo()
g.agregar_arista("Aguascalientes", "León", 120)
g.agregar_arista("Aguascalientes", "Zacatecas", 130)
g.agregar_arista("León", "Guanajuato", 55)
g.agregar_arista("León", "Querétaro", 160)
g.agregar_arista("Zacatecas", "San Luis", 190)
g.agregar_arista("Guanajuato", "Querétaro", 140)
g.agregar_arista("Querétaro", "San Luis", 200)
g.agregar_arista("San Luis", "Aguascalientes", 160)

dist, pred = g.dijkstra("Aguascalientes")
print("Distancias desde Aguascalientes:")
for ciudad, d in sorted(dist.items(), key=lambda x: x[1]):
    print(f"  {ciudad}: {d} km")

print("\nCamino a Querétaro:", g.camino(pred, "Querétaro"))
```
