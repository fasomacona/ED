## 1. Historial de navegador (dos pilas)


```python
class Navegador:
    def __init__(self):
        self.atras = []
        self.adelante = []
        self.actual = None
    
    def visitar(self, url):
        if self.actual:
            self.atras.append(self.actual)
        self.actual = url
        self.adelante.clear()
        print(f"Visitando: {url}")
    
    def ir_atras(self):
        if not self.atras:
            print("No hay historial hacia atrás")
            return
        self.adelante.append(self.actual)
        self.actual = self.atras.pop()
        print(f"Atrás → {self.actual}")
    
    def ir_adelante(self):
        if not self.adelante:
            print("No hay historial hacia adelante")
            return
        self.atras.append(self.actual)
        self.actual = self.adelante.pop()
        print(f"Adelante → {self.actual}")

nav = Navegador()
nav.visitar("google.com")
nav.visitar("youtube.com")
nav.visitar("github.com")
nav.ir_atras()
nav.ir_atras()
nav.ir_adelante()
nav.visitar("stackoverflow.com")
nav.ir_atras()
```

## 2. Cola de prioridad – salida de aviones


```python
import heapq

class ColaPrioridadAviones:
    def __init__(self):
        self._cola = []
        self._contador = 0  # para desempate por orden de llegada
    
    def encolar(self, vuelo, prioridad):
        # prioridad 1 = emergencia, 2 = regular, 3 = carga
        # heapq es min-heap, por eso usamos la prioridad directamente
        heapq.heappush(self._cola, (prioridad, self._contador, vuelo))
        self._contador += 1
        print(f"Encolado: {vuelo} (prioridad {prioridad})")
    
    def desencolar(self):
        if not self._cola:
            print("No hay aviones en espera")
            return None
        prioridad, _, vuelo = heapq.heappop(self._cola)
        print(f"Despega: {vuelo} (prioridad {prioridad})")
        return vuelo
    
    def mostrar(self):
        print("Cola actual:", [(p, v) for p, _, v in sorted(self._cola)])

aeropuerto = ColaPrioridadAviones()
aeropuerto.encolar("AA-101", 2)
aeropuerto.encolar("BB-202", 3)
aeropuerto.encolar("CC-303", 1)  # emergencia
aeropuerto.encolar("DD-404", 2)
aeropuerto.mostrar()
aeropuerto.desencolar()
aeropuerto.desencolar()
aeropuerto.mostrar()
```

## 3. Lista doblemente enlazada


```python
class NodoDoble:
    def __init__(self, dato):
        self.dato = dato
        self.anterior = None
        self.siguiente = None

class ListaDoble:
    def __init__(self):
        self.cabeza = None
        self.cola = None
    
    def insertar_inicio(self, dato):
        nuevo = NodoDoble(dato)
        if not self.cabeza:
            self.cabeza = self.cola = nuevo
        else:
            nuevo.siguiente = self.cabeza
            self.cabeza.anterior = nuevo
            self.cabeza = nuevo
    
    def insertar_final(self, dato):
        nuevo = NodoDoble(dato)
        if not self.cola:
            self.cabeza = self.cola = nuevo
        else:
            nuevo.anterior = self.cola
            self.cola.siguiente = nuevo
            self.cola = nuevo
    
    def eliminar_inicio(self):
        if not self.cabeza:
            return None
        dato = self.cabeza.dato
        self.cabeza = self.cabeza.siguiente
        if self.cabeza:
            self.cabeza.anterior = None
        else:
            self.cola = None
        return dato
    
    def eliminar_final(self):
        if not self.cola:
            return None
        dato = self.cola.dato
        self.cola = self.cola.anterior
        if self.cola:
            self.cola.siguiente = None
        else:
            self.cabeza = None
        return dato
    
    def mostrar(self):
        actual = self.cabeza
        elems = []
        while actual:
            elems.append(str(actual.dato))
            actual = actual.siguiente
        print(" ↔ ".join(elems) if elems else "Lista vacía")

ld = ListaDoble()
ld.insertar_final(10)
ld.insertar_final(20)
ld.insertar_inicio(5)
ld.insertar_final(30)
ld.mostrar()
print("Eliminado inicio:", ld.eliminar_inicio())
print("Eliminado final:", ld.eliminar_final())
ld.mostrar()
```
